## Bar plot

```python
y = [98, 67, 88, 95, 88]
x = ["part1", "part2", "part3", "part4", "part5"]
```

```python
plt.bar(x,y)               #alone this can create the graph, but shows a line "<BarContainer object of 5 artists>" if you dont want to see this line then write following line
plt.show()
```

now you can customize it by following options:

```python
color = ["red", "green", "yellow", "blue", "orange"]
plt.bar(x,y, color=color, edgecolor="black") 
plt.xlabel("parts of herry potter")
plt.ylabel("popularity", fontsize=15)
plt.title("popularity of differnet parts of harry porter")
plt.show()
```

```python
import pandas as pd 
data = pd.read_excel("expenses.xlsx")
df = pd.DataFrame(data)
print(df)
```

now i want to make a graph of amount and payment mode

```python
groupedby = df.groupby("Payment Mode")["Amount"].sum()
print(groupedby)
```

## Line Plot

```python
x = ["day1", "day2", "day3", "day4", "day5"]
y = [300, 400, 250, 230, 400]
plt.plot(x,y, marker="o")              #there are too many attribute values for marker like o, ^, *
plt.show()
```

if we want dotted line and multiple lines then

```python
y1 = [500, 450, 300, 250, 320]
plt.plot(x,y, marker="o", ls="--", color="red", label="week1")
plt.plot(x, y1, marker="^", ls="-", color="green", label="week2", alpha=0.5)          #alpha help to change opacity
plt.legend()                              #legend helps to tell the type of line
plt.show()
```

## Scatter plot

```python
import numpy as np
x = np.random.randint(1, 10, 50)             #helps to generate 50 random values b/w 1 to 10
y = np.random.randint(10, 100, 50)           #helps to generate 50 random values b/w 10 to 100
plt.scatter(x, y)
plt.show()
```

making coloured points


```python
color = np.random.randint(10, 100, 50)
plt.scatter(x,y, marker="*", c=color)
plt.colorbar()
plt.show()
```

## Pie chart

```python
brands = ["oneplus", "apple", "sumsung", "nokia", "redmi"]
x= [22,35,30,3,10]
plt.pie(x)
plt.show()
```

futher customizations ->

```python
c =["red", "pink", "blue", "green", "yellow"]
plt.pie(x, labels=brands, colors=c)
plt.show()
```

it we want to highlight redmi then we can do

```python
ex = [0,0,0.1,0,0]
plt.pie(x,labels=brands, colors=c, explode=ex, shadow=True, autopct="%.2f")    #shadow gives a shadow, autopct gives the value(#.2f means till two decimal points)
plt.show()
#we can explore more option by pressing shift + tab while writing plt.pie....
```

here we can also use excel sample file
