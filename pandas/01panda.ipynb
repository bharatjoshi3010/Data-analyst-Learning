## Working with missing data in pandas

```python
print (dataMixed.dropna())           #deletes all the rows which have any colunmn value NaN(but it can delete important data also)
```

```python
print(dataMixed.replace(np.nan, "hello"))                   #replaces all NaN value from "hello"  here np is numpy
```

for the numerical values we can replace the NaN values with the mean of the coloumn so data loss not happens and also nan value removes

```python
dataMixed["Total_Spent"] = dataMixed["Total_Spent"].replace(np.nan, dataMixed["Total_Spent"].mean())     #it changes the "total spent" nan values with its mean
```

```python
print(dataMixed)
```

```python
print(dataMixed.bfill())                        #fills the nan value same as the below cell
```

```python
print(dataMixed.ffill())                                         #fills the nan value same as the above cell
```

## Column transformation in pandas

```python
dataMixed.loc[(dataMixed["Total_Spent"] == 0), "Spent limit"] = "No spent"    #add a new column name "spent limit" and if total spent is 0 then the value of spent limit is "No spent"
dataMixed.loc[(dataMixed["Total_Spent"]>0),"Spent limit"] = "spent good"  
print(dataMixed)
```

## Merge, Join, Concatenate :

```python
data1 = {"Emp Id": ["E01", "E02", "E03", "E04", "E05", "E06"],
        "Names": ["Ram", "Shyam", "Rahul", "Vishal", "ravi", "John"],
        "Age": [34, 56,25,44,32,36]
        }
data2 = {"Emp Id": ["E01", "E03", "E06", "E04", "E09", "E08"],
        "Salary": [45000,56000,34000,30000,50000,62000]}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print(df1)
print()
print(df2)
```

lets merge it

```python
print(pd.merge(df1, df2, on = "Emp Id"))             #it merge only those rows which have same "Emp Id" in both rows.(inner join)
```

```python
print(pd.merge(left=df1, right=df2, on="Emp Id", how="left"))     #left join
```

```python
print(pd.merge(left=df1, right=df2, on="Emp Id", how="right"))          #right join
```

```python
print(pd.join(df1, df2))
```

```python
print(pd.concat([df1,df2]))                              #if both table have different data then it concatinate
```

## Comapre data frames in python

```python
dict = {"Fruits" : ["mango", "apples", "banana", "papaya"],
       "Price": [100,150,50,35],
       "Quantity": [15,10,10,3]}
df1 = pd.DataFrame(dict)
print(df1)
df2 = df1.copy()
print(df2)
```

now boht the tables have same data but we want to compare it so there should be some difference in them

```python
df2.loc[0, "Price"] = 120
df2.loc[1, "Price"] = 180
df2.loc[3, "Price"] = 37
df2.loc[0, "Quantity"] = 12
df2.loc[1, "Quantity"] = 16
df2.loc[3, "Quantity"] = 7
print(df2)
```

now both the table are different so we can compare it

```python
print(df1.compare(df2))             #self -> df1 value, other -> df2 value
```

```python
print(df1.compare(df2, align_axis=0))                    #by doing this we get the result horizentaly
```

```python
print(df1.compare(df2, keep_shape=True))                       #show Nan for those values which are same
```

## Pivoting and Melting dataframes

```python
dict = {"keys":["k1", "k2", "k1", "k2"],
        "names": ["John", "Ben", "David", "Peter"],
       "houses": ["red", "blue", "green", "red"],
       "Grades": ["3rd", "8th", "9th", "8th"]}
df = pd.DataFrame(dict)
print(df)
print(df.pivot(index="keys", columns="names", values=["houses", "Grades"]))         #names are unique so we make then as colums , and make keys as index
```

```python
print(pd.melt(df, id_vars=["names"], value_vars=["houses", "Grades"], var_name="Houses&Grades", value_name="valueH&G"))
```

```python
print(pd.melt(df, id_vars=["names"], value_vars=["houses"]))          #it converts the "houses" column to a variable
```
